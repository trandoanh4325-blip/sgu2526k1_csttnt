### 1. Đồ thị

```python
ADJ = {
    'A' : ['B', 'C', 'F'],
    'B' : ['A', 'C', 'E', 'F'],
    'C' : ['A', 'B', 'D', 'F'],
    'D' : ['C', 'F'],
    'E' : ['B', 'D', 'F'],
    'F' : ['A', 'C', 'E']
}
```

```python
import pprint
pprint.pprint(ADJ)
```

#### Đường đi từ A -> D

##### DFS

```python
def DFS(graph, start, goal, path=None, visited=None):
    if path is None:
        path = []
    if visited is None:
        visited = set()

    path.append(start)
    visited.add(start)

    if start == goal:
        return path

    for neighbor in graph[start]:
        if neighbor not in visited:
            result = DFS(graph, neighbor, goal, path.copy(), visited.copy())
            if result:
                return result

    return []

dfs = DFS(ADJ, 'A', 'D')
print("Đường đi từ A đến D:", " -> ".join(dfs))

```

##### BFS

```python
from collections import deque
def BFS(graph, start, goal):
    queue = deque([[start]])
    visited = set()

    while queue:
        path = queue.popleft()
        node = path[-1]

        if node == goal:
            return path

        if node not in visited:
            visited.add(node)
            for neighbor in graph[node]:
                if neighbor not in visited:
                    queue.append(path + [neighbor])

    return []

bfs = BFS(ADJ, 'A', 'D')
print("Đường đi từ A đến D:", " -> ".join(bfs))
```

#### Bậc của từng đỉnh đồ thị, bậc cao nhất và vẽ đồ thị cho bên trên.

```python
# Tính bậc của từng đỉnh
bac = {}
for v in ADJ:
    bac[v] = len(ADJ[v])

print("Bậc của từng đỉnh:")
for v in sorted(bac):
    print(f"Dinh {v}: bac = {bac[v]}")

# Bậc cao nhất
max_bac = max(bac.values())
max_dinh = [v for v in bac if bac[v] == max_bac]

print("\nBậc cao nhất của đồ thị:", max_bac)
print("Đỉnh có bậc cao nhất:", ", ".join(max_dinh))
```

### 2.Mê cung

```python
from collections import deque

# kích thước mê cung
N = 6
START = (0, 0)
GOAL  = (5, 5)

# wall[r][c] = [up, right, down, left]
# 1 = có tường, 0 = không có tường
wall = [[[0,0,0,0] for _ in range(N)] for _ in range(N)]

# KHAI BÁO TƯỜNG
# tường dọc giữa cột 2 và 3 hàng 4 và 5
for r in [0,1,3,4]:
    wall[r][1][1] = 1   
    wall[r][2][3] = 1     

# tường dọc giữa cột 2 và 3 hàng 4 và 5
for r in [3, 4]:
    wall[r][1][1] = 1   
    wall[r][2][3] = 1 
    
# tường ngang giữa hàng 3 và 4
for c in [4,5]:
    wall[2][c][2] = 1     
    wall[3][c][0] = 1     

```

```python
def bfs():
    q = deque([START])
    parent = {START: None}

    while q:
        r, c = q.popleft()
        if (r, c) == GOAL:
            break

        # up
        if r > 0 and wall[r][c][0] == 0 and (r-1, c) not in parent:
            parent[(r-1, c)] = (r, c)
            q.append((r-1, c))

        # right
        if c < N-1 and wall[r][c][1] == 0 and (r, c+1) not in parent:
            parent[(r, c+1)] = (r, c)
            q.append((r, c+1))

        # down
        if r < N-1 and wall[r][c][2] == 0 and (r+1, c) not in parent:
            parent[(r+1, c)] = (r, c)
            q.append((r+1, c))

        # left
        if c > 0 and wall[r][c][3] == 0 and (r, c-1) not in parent:
            parent[(r, c-1)] = (r, c)
            q.append((r, c-1))

    # truy vết đường đi
    path = []
    cur = GOAL
    while cur is not None:
        path.append(cur)
        cur = parent[cur]

    return path[::-1]
```

```python

path = bfs()
print("Đường đi của kiến từ đến A:")
for step in path:
    print(step)
```

### 3.Trò chơi 8-Puzzles:

```python
from collections import deque
import random
def ngaunhien():
    while True:
        board = list(range(9))
        random.shuffle(board)

        inv = 0
        for i in range(9):
            for j in range(i + 1, 9):
                if board[i] != 0 and board[j] != 0 and board[i] > board[j]:
                    inv += 1

        if inv % 2 == 0:
            return tuple(board)

start = ngaunhien()

goal = (1, 2, 3,
        4, 5, 6,
        7, 8, 0)
```

```python
# các vị trí có thể đổi với ô trống
moves = {
    0: [1, 3],
    1: [0, 2, 4],
    2: [1, 5],
    3: [0, 4, 6],
    4: [1, 3, 5, 7],
    5: [2, 4, 8],
    6: [3, 7],
    7: [4, 6, 8],
    8: [5, 7]
}
```

```python
def bfs(start, goal):
    q = deque([start])
    parent = {start: None}

    while q:
        state = q.popleft()

        if state == goal:
            break

        zero = state.index(0)

        for nxt in moves[zero]:
            new_state = list(state)
            new_state[zero], new_state[nxt] = new_state[nxt], new_state[zero]
            new_state = tuple(new_state)

            if new_state not in parent:
                parent[new_state] = state
                q.append(new_state)

    # truy vết đường đi
    path = []
    cur = goal
    while cur is not None:
        path.append(cur)
        cur = parent[cur]

    return path[::-1]
```

```python
path = bfs(start, goal)

print("Trạng thái ban đầu:")
print(start[0:3])
print(start[3:6])
print(start[6:9])

print("\nTrạng thái sau khi xếp xong:")
print(goal[0:3])
print(goal[3:6])
print(goal[6:9])
```
